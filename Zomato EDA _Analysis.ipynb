```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# setting up visualisation style
plt.style.use('seaborn-v0_8')
```

```python
# Load the dataset
df = pd.read_csv("zomato_.csv", encoding="latin-1")   # encoding = latin-1 is used for read special character

# Display firwst 5 rows to understand the data
df.head()
```

```python
# check the shape of thge dataset
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# checking the datatypes of columns
df.info()
```

```python
# Display summary statistics for numerical columns
df.describe(include="all")
```

```python
# check for missing values
df.isnull().sum()

# Here we check this to identify incomplete data that may affect analysis
```

```python
# check duplicated
print("Duplicate Rows:", df.duplicated().sum())

# If there is any removing it
df = df.drop_duplicates()
print("After Removing Duplicates:", df.duplicated().sum())
```

```python
# check unique countries and location
df['Country Code'].unique()

# to know from which countries the zomato data has been collected
```

```python
# Most common cuisines
top_cuisines = df['Cuisines'].value_counts().head(10)

plt.figure(figsize=(8,5))
sns.barplot(x=top_cuisines.values, y=top_cuisines.index, palette='viridis', ec="k")
plt.title("Top 10 Most Popular Cuisines")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine Type")
plt.show()

# Reason
# Helps to understand which cuisines are most preferred by customers
```

```python
# Distribution of Aggregate Rating
plt.figure(figsize=(8,5))
sns.histplot(df["Aggregate rating"], bins=20, kde=True, color="orange")
plt.title("Distribution of Restaurnat Rating")
plt.xlabel("Aggregate Rating")
plt.ylabel("Count")
plt.show()

# Reason
# To check how customer rating are spread across Restaurants
```

```python
# Average rating by city
avg_rating_city = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=avg_rating_city.values, y=avg_rating_city.index, palette = 'coolwarm')
plt.title("Top 10 Cities by Average Rating")
plt.xlabel("Average Rating")
plt.ylabel("City")
plt.show()

# Reason
# Helps identify which cities have better-rated restaurants
```
